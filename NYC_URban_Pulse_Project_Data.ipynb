```python
api_url = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"

row_limit = 100000
batch_size = 500

all_records = []
```

```python
for offset in range(0, row_limit, batch_size):
    parameters = {
        "$select": "unique_key,created_date,agency,agency_name,complaint_type,descriptor,location_type,incident_zip,incident_address,street_name,address_type,city,facility_type,status,due_date,resolution_description,resolution_action_updated_date,community_board,council_district,police_precinct,bbl,borough,open_data_channel_type,latitude,longitude",
        "$limit": batch_size,
        "$offset": offset,
        "$order": "created_date DESC"
    }

    response = requests.get(api_url, params=parameters, timeout=200)

    print("HTTP status:", response.status_code)
    response.raise_for_status()

    batch_records = response.json()
    all_records.extend(batch_records)

    print("Records collected so far:", len(all_records))

    if len(batch_records) < batch_size:
        print("No more records returned.")
        break
```

```python
nyc311_df = pd.DataFrame(all_records)

print("Rows collected:", len(nyc311_df))
print("Columns collected:", len(nyc311_df.columns))

nyc311_df.head()
```

```python
print("Column names:")
print(nyc311_df.columns.tolist())

print("\nData types:")
print(nyc311_df.dtypes)
```
